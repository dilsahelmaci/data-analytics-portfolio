### **5- Data Wrangling and Preparation**
---

```python
df.info()
```

#### 5.1- Format `price` column and handling missing values

```python
# Convert 'price' column to numeric format, removing non-numeric characters in it
df['price'] = df['price'].str.replace('$', '', regex=False).str.replace(',', '', regex=False).astype(float)

# Drop rows with missing 'price' values
df = df.dropna(subset=['price'])

# Check price stats
print(df['price'].describe())
```

#### 5.2- Convert `host_since` column to datetime

```python
#  Convert `host_since` to datetime, so missing dates are NaT
df['host_since'] = pd.to_datetime(df['host_since'], errors='coerce')
print(f"`host_since` dtype after conversion: {df['host_since'].dtype}")
```

#### 5.3- Impute missing `host_is_superhost` as False (not superhost)

```python
# Fill missing 'host_is_superhost' column with 'f'
df['host_is_superhost'] = df['host_is_superhost'].fillna('f').astype('category')
df['host_is_superhost'].value_counts()
```

#### 5.4- Impute missing `review_scores_rating` with median

```python
# Fill missing 'review_score_rating' column with the median value to maintain the distribution's central tendency
df['review_scores_rating'] = df['review_scores_rating'].fillna(df['review_scores_rating'].median())
```

```python
df.info()
```

#### 5.5- Convert `room_type`, `neighbourhood_cleansed`, and `neighbourhood_group_cleansed` columns to pandas category dtype

```python
# Convert categorical columns to category dtype
for col in ['room_type', 'neighbourhood_cleansed', 'neighbourhood_group_cleansed']:
    df[col] = df[col].astype('category')
```

```python
df.info()
```

### **6- Feature Engineering**
---

#### 6.1- Compute host experience in years

```python
# Compute experience in years (NaT values will yield NaN)
df['host_experience_years'] = (pd.Timestamp.today() - df['host_since']).dt.days / 365

# Fill missing experience with 0
df['host_experience_years'] = df['host_experience_years'].fillna(0)

# Show `host_experience_year` stats after cleaning
df['host_experience_years'].describe()
```

#### 6-2. Parse amenities into a list and count

```python
# Parse amenities list from string, removing braces and quotes
df['amenities_list'] = df['amenities'].str.strip('[]').str.replace('"', '').str.split(',')

# Standardize amenity names (lowercase, strip spaces)
df['amenities_list'] = df['amenities_list'].apply(lambda x: [a.strip().lower() for a in x])

# Count number of amenities for each listing
df['amenities_count'] = df['amenities_list'].apply(len)
```

```python
df.isnull().sum()
```

**Note on `host_since` missing values:** 

All columns, except `host_since`, are complete with no missing values.  
This is not a problem for our analysis, as we rely on the derived feature `host_experience_years` (years since host joined) for EDA.  
For listings with a missing `host_since` date, `host_experience_years` is set to 0, ensuring no gaps in downstream analyses.  
We retain the original `host_since` column for transparency and to enable any future analyses on host tenure.
